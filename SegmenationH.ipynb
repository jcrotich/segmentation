# Segmentation model Option 1
K-Means is simple, fast, and highly effective for partitioning large datasets like the shared into k distinct clusters. It works well for the clusters are compact, spherical, and evenly sized, often assumed in segmentation tasks.

considering the standardized numerical data (using StandardScaler or MinMaxScaler) and categorical data encoded (LabelEncoder), K-Means performs well for segmentation

# Assumptions of K-Means
1. Cluster Shape: Assumes clusters are spherical/round in shape and equally sized.
2. Data Scaling: Requires all features to be on a similar scale to ensure fair weighting in distance calculations.(Euclidean distance suitability)
3. Feature Independence: Assumes features are independent. Strongly correlated features can bias the clustering process.(Equal variance)
4. Cluster Membership: Every data point belongs to one and only one cluster.
5. Predefined Number of Clusters (k): The number of clusters (k) must be determined beforehand using techniques like the Elbow Method or Silhouette Scores.

# Advantages to using K-Means
1. Scalability: K-Means is computationally efficient for large datasets.
2. Interpretability: Results are easy to understand and visualize.
3. Broad applicability: Works well on numerical data after preprocessing

# Limitations:

1. K-Means can easily struggle with irregularly shaped clusters.
2. Outliers can skew the cluster centers.
3. The number of clusters k must be predefined.

### Option 2
#  Gaussian Mixture Model (GMM)

it is a probabilistic model that allows soft clustering, where, each data point has a probability of belonging to multiple clusters, useful if clusters have overlapping boundaries.

Ideal for complex datasets where clusters may overlap, and probabilistic assignments are needed. It is a natural extension of K-Means.

# Assumptions of Gaussian Mixture Models (GMM)

1. GMM assumes that the data points in each cluster are generated from a Gaussian (normal) distribution, having its own mean and covariance matrix that define the distribution.

2. Unlike K-Means, GMM allows clusters to overlap by assigning probabilities of belonging to multiple clusters.

3. GMM assumes that each data point can belong to multiple clusters, but with different probabilities.

4. GMM can handle non-spherical/elliptical cluster shapes due to its flexibility in fitting the Gaussian distributions.

5. The features are assumed to be independent within each Gaussian distribution.


# Advantages of using GMM

1. GMM is flexible in that handles elliptical and non-spherical clusters better than K-Means.
2. It provides probabilities for cluster membership, giving more nuanced results.
3. GMM Assumes data is generated from Gaussian distributions, which can be realistic in many cases.

# Limitations of GMM

1. GMM can be slower for large datasets compared to K-Means.
2. Results may vary based on initialization.
3. Choosing the correct number of clusters remains a challenge.




### Option 3

# DBSCAN (Density-Based Spatial Clustering)

It groups data into dense clusters and identifies noise/outliers, highly effective when clusters are non-spherical and when there is noise in the dataset.
Works best when clusters are not spherical, and you want to detect noise. It may however require subsampling of the dataset or careful parameter tuning

# Assumptions of DBSCAN (Density-Based Spatial Clustering)

1. DBSCAN assumes that clusters are regions of high density separated by regions of low density (sparse points).

2. Clustering depends on two key parameters:
eps (ε): The maximum radius of the neighborhood around a point.
min_samples: The minimum number of points required to form a dense region.
These parameters must be well-tuned based on the data distribution.

3. Unlike K-Means, DBSCAN does not assume spherical clusters and can detect complex shapes.

4. Points that do not belong to any dense region are classified as noise.

5. DBSCAN assumes that a distance metric (like Euclidean distance) is appropriate for measuring the density of points.

# Advantages of DBSCAN

1. Automatically identifies the number of clusters(k).
2. Handles noisy data and identifies outliers as "noise" points.
3. Works well for arbitrarily shaped clusters.

# Limitations of DBSCAN

1. It can be slow for very large datasets.
2. Requires careful tuning of eps (radius) and min_samples.
3. Struggles if clusters have varying densities.

# Model Evaluation
1. Using Silhouette Score which measures how similar a point is to its own cluster (cohesion) compared to other clusters (separation).

The evalaution is best for K-Means and GMM,  DBSCAN leaves out noise points (which complicates evaluation

Range: [-1, 1]
- Values close to 1: Good clustering/ points are well-clustered.
- Values close to 0: Clusters overlap significantly.
- Values close to -1: Poor clustering / misclassified points

2.  Model-Specific Metrics( Inertia for K-Means, BIC/AIC for GMM, Noise Detection for DBSCAN)
- K-Means: Evaluate using inertia (sum of squared distances to cluster centers). Lower inertia indicates better clustering. Elbow Method helps determine the optimal number of clusters

- Gaussian Mixture Model (GMM) Use Log-Likelihood: GMM optimizes the likelihood of data under the Gaussian distributions.
Bayesian Information Criterion (BIC) or Akaike Information Criterion (AIC): These are used to determine the optimal number of components (clusters). Lower values indicate better models.

- DBSCAN Noise Detection (number of points labeled as noise), to evaluate the proportion of outliers in the dataset.
